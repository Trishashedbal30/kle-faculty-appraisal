# Lab 1 — Clarify, plan and revise

This notebook drafts a plan only. It does not score or release an appraisal. Open it in VS Code with the `.venv` interpreter.


```python
from pathlib import Path
import sys, json
project = Path.cwd() if (Path.cwd() / 'app').exists() else Path.cwd().parent
sys.path.insert(0, str(project))
from app.contracts import PlanInput
from app.skills import plan_skill
```

```python
request = input('What appraisal would you like to prepare? ')
questions = ['Which faculty ID and cycle should the plan cover?', 'Which approved institutional policy and track apply?', 'Are research, teaching and service records complete?']
for question in questions:
    print(question)
```

```python
# Planning exercise only: policy is a placeholder, not a scoring authority.
faculty_id = input('Faculty ID: ')
policy_id = input('Approved policy ID (from the application): ')
profile = input('Track [balanced/teaching/research/administrative]: ') or 'balanced'
completeness = {area: input(area + ' completeness declaration: ') for area in ('research', 'teaching', 'administrative')}
data = PlanInput(faculty_id=faculty_id, policy_id=policy_id, profile=profile, request=request, completeness=completeness)
preview = plan_skill(data, 'NOTEBOOK_PREVIEW_ONLY', {'id': policy_id})
print(json.dumps(preview, indent=2))
```

```python
feedback = input('Revise the request, or press Enter to retain it: ')
if feedback.strip():
    data = PlanInput(**{**data.model_dump(), 'request': feedback})
    preview = plan_skill(data, 'NOTEBOOK_PREVIEW_ONLY', {'id': policy_id})
print(json.dumps(preview, indent=2))
print('Use Appraisal workspace to create and lock the actual authenticated plan.')
```

## Acceptance

A vague request produces at least two clarifying questions. The output is an editable structured plan. Revision produces a new plan rather than pretending to return a final report. Actual policy, consent and identity checks run in the authenticated application.
